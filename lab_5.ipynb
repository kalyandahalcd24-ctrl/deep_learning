```python

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import layers, models

np.random.seed(42)
tf.random.set_seed(42)

N = 1000
t = np.arange(N)
series = np.sin(2 * np.pi * t / 50) + 0.01 * t + 0.1 * np.random.randn(N)

SPLIT = int(0.8 * N)

# Task 1 - Scale using TRAIN-ONLY statistics
mn = np.min(series[:SPLIT])
mx = np.max(series[:SPLIT])
scaled = (series - mn) / (mx - mn)

# Task 2 - Build sliding windows
W = 30

def make_windows(a, W):
    X, y = [], []
    for i in range(len(a) - W):
        X.append(a[i:i + W])
        y.append(a[i + W])
    return np.array(X), np.array(y)

X, y = make_windows(scaled, W)

# Task 3 - Shape and split in time order
X = X.reshape((X.shape[0], W, 1))

TR = SPLIT - W

X_train, y_train = X[:TR], y[:TR]
X_test, y_test = X[TR:], y[TR:]

# Task 4 - LSTM Regressor
model = models.Sequential()

model.add(layers.LSTM(50, input_shape=(W, 1)))
model.add(layers.Dense(1))

model.compile(
    optimizer='adam',
    loss='mse',
    metrics=['mae']
)

# Task 5 - Train and score
history = model.fit(
    X_train,
    y_train,
    epochs=20,
    batch_size=32,
    validation_data=(X_test, y_test)
)

pred = model.predict(X_test).ravel()

rmse = np.sqrt(np.mean((pred - y_test) ** 2))

print("Test RMSE:", rmse)
```
